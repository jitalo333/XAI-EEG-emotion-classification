# Method 1 (Shap)

This method applies SHAP (SHapley Additive exPlanations) to interpret the model's predictions. It uses a KernelExplainer to approximate the importance of each EEG feature (channels and frequency bands) by perturbing the input data.

The method expects 3D tensors with the shape (Samples, Channels, Bands)
Processing time depends on nsamples. A GPU is recommended for values > 100.

background_data: A subset of training data used as the reference baseline (numpy array).
eval_data: The specific samples you want to explain (numpy array).
class_idx: The target class index to analyze (e.g., 0 for Negative, 1 for Positive).
nsamples: Number of perturbation samples (default: 100). Higher values yield more accurate results but increase runtime.

This demonstration uses randomly generated data to showcase how the code structure works without requiring external datasets.

```python
import numpy as np
from XAI import XAI_module

model_path = "AI_models/weights/CDCN_SEED_dependent.pth"
savepath = "results/shap_analysis/"
verbose = True
explainable = XAI_module(model_path, savepath, verbose)

n_samples = 50   
n_bg = 100       
n_channels = 62  
n_bands = 5   

# Datos aleatorios simulando EEG
train_data_mock = np.random.rand(n_bg, n_channels, n_bands).astype(np.float32)
eval_data_mock = np.random.rand(n_samples, n_channels, n_bands).astype(np.float32)

args = {
    'background_data': train_data_mock, 
    'eval_data': eval_data_mock,
    'class_idx': 1,  
    'nsamples': 100
    }
explainable.method1(args)
```

# Method 2 (Lime)

This method applies LIME to explain individual model predictions. Unlike SHAP, which can explain an entire dataset, LIME approximates the neural network locally around a single instance using a simple, interpretable linear model. It identifies which channel-frequency combinations (e.g., FP1_Alpha) were most influential for a specific classification.

Input Format: The method expects 3D tensors. The training data must be (Samples, Channels, Bands) and the instance to explain must be (1, Channels, Bands).

Output: An interactive HTML report.

train_data: The training dataset (numpy array). LIME uses this to learn the feature distribution and discretize values.
eval_instance: The single EEG sample you want to explain (numpy array with shape 1, C, B).
class_names: List of class labels (e.g., ['Negative', 'Neutral', 'Positive']).
num_features: Number of top contributing features to show in the report (default: 15).

The code below generates a random tensor to simulate a single EEG trial (eval_instance). Since this input contains no real physiological patterns, the resulting feature importance list and probability scores are purely technical demonstrations and should not be interpreted biologically.

```python
import numpy as np
from XAI import XAI_module

model_path = "AI_models/weights/CDCN_SEED_dependent.pth"
savepath = "results/lime_analysis/"
verbose = True
explainable = XAI_module(model_path, savepath, verbose)

n_bg = 100
n_channels = 62 
n_bands = 5

train_data_mock = np.random.rand(n_bg, n_channels, n_bands).astype(np.float32)
single_instance = np.random.rand(1, n_channels, n_bands).astype(np.float32)
args = {
    'train_data': train_data_mock, 
    'eval_instance': single_instance,
    'class_names': ['Negative', 'Neutral', 'Positive'],
    'num_features': 10
    }
explainable.method2(args)
```
